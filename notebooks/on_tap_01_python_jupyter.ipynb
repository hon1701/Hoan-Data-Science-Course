```python
students = pd.DataFrame({
    "name": ["An", "Binh", "Chi", "Dung", "Hoa"],
    "age": [20, 22, 19, 21, 23],
    "score": [7.0, 9.0, 6.5, 8.5, 9.5]
})
print(students)
print(students.shape)
print(students["score"])
print(students.iloc[2])
print(students[students["score"] >= 8])
print(students[(students["score"] >= 8) & (students["age"]>=21)])
students['passed'] = students['score']>=5
print(students)
print(students.sort_values("score", ascending=False))
print(students.sort_values("score", ascending=False).head(2))
```
