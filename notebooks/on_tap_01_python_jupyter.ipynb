```python
x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(x[1, 1:])
print(x[:, 1])
print(x[x > 50])
```

x[1, 1:]
  ↑  ↑
  │  └─ từ cột 1 trở đi
  └──── hàng 1

x[:, 1]
  ↑  ↑
  │  └─ cột 1
  └──── tất cả hàng

x[x > 50]
   ↑
boolean mask
