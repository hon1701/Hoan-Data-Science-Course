Câu 6: Giải thích sự khác nhau giữa keep_default_na=True và keep_default_na=False

keep_default_na=True
→ sử dụng danh sách giá trị NA mặc định của Pandas.
→ nếu có thêm na_values, các giá trị đó được bổ sung.

keep_default_na=False
→ không sử dụng danh sách NA mặc định.
→ nếu có na_values, chỉ những giá trị mình chỉ định
  mới được coi là missing.

```python
chunker = pd.read_csv(
    path,
    chunksize=1000
)

tot = pd.Series([], dtype="int64")

for piece in chunker:
    counts = piece["key"].value_counts()

    tot = tot.add(
        counts,
        fill_value=0
    )

tot = tot.sort_values(
    ascending=False
)

tot
```
